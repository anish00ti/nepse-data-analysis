```python
import random
from datetime import datetime, timedelta

def generate_market_data(days=7):
    """Generates synthetic daily trading metrics."""
    base_price = 2100.0
    records = []
    
    current_date = datetime.now() - timedelta(days=days)
    
    for i in range(days):
        date_str = (current_date + timedelta(days=i)).strftime("%Y-%m-%d")
        price_change = round(random.uniform(-35.0, 40.0), 2)
        base_price = max(1000.0, round(base_price + price_change, 2))
        volume = random.randint(150000, 850000)
        
        records.append({
            "date": date_str,
            "closing_index": base_price,
            "volume": volume,
            "status": random.choice(["BULLISH", "BEARISH", "NEUTRAL"])
        })
    return records

def calculate_summary(data):
    """Calculates key summary statistics from dataset."""
    indices = [d["closing_index"] for d in data]
    total_volume = sum(d["volume"] for d in data)
    
    avg_index = round(sum(indices) / len(indices), 2)
    min_index = min(indices)
    max_index = max(indices)
    
    return {
        "avg": avg_index,
        "min": min_index,
        "max": max_index,
        "total_vol": total_volume
    }

def main():
    print("=" * 45)
    print("      AUTOMATED DATA SUMMARY REPORT          ")
    print("=" * 45)
    
    dataset = generate_market_data(days=5)
    stats = calculate_summary(dataset)
    
    print("\nDaily Breakdown:")
    print(f"{'Date':<12} | {'Index':<10} | {'Volume':<10} | {'Trend'}")
    print("-" * 45)
    for row in dataset:
        print(f"{row['date']}  | {row['closing_index']:<10} | {row['volume']:<10} | {row['status']}")
        
    print("-" * 45)
    print("\nPeriod Metrics:")
    print(f"• Average Index : {stats['avg']}")
    print(f"• Highest Index : {stats['max']}")
    print(f"• Lowest Index  : {stats['min']}")
    print(f"• Total Volume  : {stats['total_vol']:,} units")
    print("=" * 45)

if __name__ == "__main__":
    main()
```
