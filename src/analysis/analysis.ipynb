```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from datetime import date
import os
```

```python
from pathlib import Path

today_date = date.today().isoformat()
candidate_roots = [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]
file_path = None

for root in candidate_roots:
    candidate = root / "data" / "raw" / "stock_data" / f"{today_date}-stock_data.csv"
    if candidate.exists():
        file_path = candidate
        break

if file_path is None:
    raise FileNotFoundError(f"Stock data file not found for {today_date} in data/raw/stock_data")

df = pd.read_csv(file_path)
df
```
