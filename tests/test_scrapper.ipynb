```python
import requests
import re

url = "https://www.nepalstock.com/floor-sheet"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers, timeout=30)

print("Status:", response.status_code)

# Find JavaScript files
js_files = re.findall(
    r'<script[^>]+src="([^"]+\.js)"',
    response.text
)

print("JavaScript files:")

for js in js_files:
    print(js)
```

```python
from urllib.parse import urljoin

# Find main JavaScript file
main_js = next(
    (js for js in js_files if "main." in js),
    None
)

print("\nMain JS:", main_js)

if main_js:
    js_url = urljoin(url, main_js)

    js_response = requests.get(
        js_url,
        headers=headers,
        timeout=30
    )

    print("JS Status:", js_response.status_code)
    print("JS Length:", len(js_response.text))

    # Search for floor-sheet related text
    matches = re.findall(
        r'[^"\']*(?:floor-sheet|floorsheet|floorSheet)[^"\']*',
        js_response.text,
        re.IGNORECASE
    )

    print("\nFloor-sheet matches:")

    for match in matches[:50]:
        print(match)
```

```python
# Find the position of getAllFloorSheet in the JavaScript
position = js_response.text.find("getAllFloorSheet")

print("Position:", position)

# Print a larger section around it
if position != -1:
    print(js_response.text[position:position + 3000])
else:
    print("❌ getAllFloorSheet not found")
```

```python
js = js_response.text

keywords = [
    "nots/nepse-data/floorsheet",
    "nepse-data/floorsheet",
    "getAllFloorSheet"
]

for keyword in keywords:
    position = js.find(keyword)

    print("\n" + "=" * 80)
    print("KEYWORD:", keyword)
    print("POSITION:", position)

    if position != -1:
        print(js[position - 500:position + 2000])
```

```python
import requests
import json

api_url = "https://www.nepalstock.com/api/nots/nepse-data/floorsheet"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Content-Type": "application/json",
    "Accept": "application/json, text/plain, */*"
}

payload = {
    "id": 1
}

params = {
    "page": 0,
    "sort": "contractId,desc"
}

response = requests.post(
    api_url,
    headers=headers,
    params=params,
    json=payload,
    timeout=30
)

print("Status:", response.status_code)
print("URL:", response.url)
print("Response length:", len(response.text))

print("\nResponse:")
print(response.text[:2000])
```

```python
# Search for common API base URL patterns
patterns = [
    r'https?://[^"\']+',
    r'https?://[^"\']*api[^"\']*'
]

for pattern in patterns:
    matches = re.findall(pattern, js_response.text, re.IGNORECASE)

    print("\nMatches:")
    for match in matches[:100]:
        print(match)
```
